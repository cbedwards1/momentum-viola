## 1. Define the system and its direction

Our system contains **Cart 1 and Cart 2**.

We choose a sign convention:

- Rightward velocity is positive.
- Leftward velocity is negative.
- A cart at rest has zero velocity.

Mass is measured in kilograms and velocity is measured in meters per second.

The starting values below match the default example in the Voilà application.

```python
# -----------------------------
# STUDENT-EDITABLE INPUTS
# -----------------------------

# Cart 1 starts on the left.
m1 = 1.0       # Cart 1 mass in kilograms
u1 = 2.0       # Cart 1 initial velocity in meters per second

# Cart 2 starts on the right.
m2 = 1.0       # Cart 2 mass in kilograms
u2 = 0.0       # Cart 2 initial velocity in meters per second

# Coefficient of restitution:
# e = 1.0 means elastic
# e = 0.0 means perfectly inelastic
# values between 0 and 1 mean partially inelastic
e = 1.0

print("Inputs loaded.")
print(f"Cart 1: mass = {m1:.2f} kg, initial velocity = {u1:+.2f} m/s")
print(f"Cart 2: mass = {m2:.2f} kg, initial velocity = {u2:+.2f} m/s")
print(f"Coefficient of restitution: e = {e:.2f}")
```

## 2. Calculate initial momentum

Linear momentum is a vector quantity:

\[
p=mv
\]

In Python, multiplication is written with an asterisk: `*`.

Read the next cell carefully. Find the two lines where the model multiplies mass by velocity.

```python
# Individual initial momenta
p1_i = m1 * u1
p2_i = m2 * u2

# Total initial momentum of the two-cart system
p_total_i = p1_i + p2_i

print(f"Cart 1 initial momentum: {p1_i:+.3f} kg·m/s")
print(f"Cart 2 initial momentum: {p2_i:+.3f} kg·m/s")
print(f"Total initial system momentum: {p_total_i:+.3f} kg·m/s")
```

## 3. Why can momentum be negative?

Mass is positive in this model. Velocity can be positive, zero, or negative because velocity includes direction.

Therefore, the sign of momentum follows the sign of velocity:

- Positive velocity gives positive momentum.
- Zero velocity gives zero momentum.
- Negative velocity gives negative momentum.

A negative momentum does **not** mean the cart has “less than no momentum.” It means the momentum points in the direction we defined as negative—leftward in this model.

```python
# Small sign experiment
test_mass = 2.0
test_velocity = -3.0
test_momentum = test_mass * test_velocity

print(f"Test mass: {test_mass:.1f} kg")
print(f"Test velocity: {test_velocity:+.1f} m/s")
print(f"Test momentum: {test_momentum:+.1f} kg·m/s")
```

## 4. Calculate the collision outcome

The full application uses conservation of momentum together with a coefficient of restitution.

The coefficient of restitution, `e`, describes how the carts separate after the collision:

- `e = 1.0`: elastic collision
- `e = 0.0`: perfectly inelastic collision
- `0 < e < 1`: partially inelastic collision

The equations below calculate the two final velocities. You are not expected to memorize the Python expression. Focus on how the inputs `m1`, `u1`, `m2`, `u2`, and `e` are used to generate `v1` and `v2`.

```python
# Check whether Cart 1 can catch Cart 2.
# Cart 1 starts to the left, so a collision requires u1 > u2.
if u1 <= u2:
    raise ValueError(
        "These inputs do not produce a collision: Cart 1 must have "
        "a greater initial velocity than Cart 2."
    )

total_mass = m1 + m2
relative_approach_speed = u1 - u2

# Final velocity of Cart 1
v1 = (
    m1 * u1
    + m2 * u2
    - m2 * e * relative_approach_speed
) / total_mass

# Final velocity of Cart 2
v2 = (
    m1 * u1
    + m2 * u2
    + m1 * e * relative_approach_speed
) / total_mass

print(f"Cart 1 final velocity: {v1:+.3f} m/s")
print(f"Cart 2 final velocity: {v2:+.3f} m/s")
```

## 5. Test momentum conservation

The model now repeats the same momentum operation with the **final** velocities.

Look again for mass multiplied by velocity. Then locate the line that adds the two final momenta.

```python
# Individual final momenta
p1_f = m1 * v1
p2_f = m2 * v2

# Total final momentum of the two-cart system
p_total_f = p1_f + p2_f

# Compare the system totals
change_in_system_momentum = p_total_f - p_total_i

print(f"Cart 1 final momentum: {p1_f:+.3f} kg·m/s")
print(f"Cart 2 final momentum: {p2_f:+.3f} kg·m/s")
print(f"Total initial momentum: {p_total_i:+.3f} kg·m/s")
print(f"Total final momentum:   {p_total_f:+.3f} kg·m/s")
print(f"Change in system momentum: {change_in_system_momentum:+.6f} kg·m/s")
```

## 6. Compare momentum with kinetic energy

Momentum is conserved in the isolated model for every collision type. Kinetic energy is conserved only for an elastic collision.

The kinetic-energy equation is:

\[
KE=\frac{1}{2}mv^2
\]

Because velocity is squared, kinetic energy does not become negative when a cart moves left.

```python
ke_i = 0.5 * m1 * u1**2 + 0.5 * m2 * u2**2
ke_f = 0.5 * m1 * v1**2 + 0.5 * m2 * v2**2
change_in_ke = ke_f - ke_i

print(f"Initial system kinetic energy: {ke_i:.3f} J")
print(f"Final system kinetic energy:   {ke_f:.3f} J")
print(f"Change in kinetic energy:      {change_in_ke:+.3f} J")
```

## 8. Inspect the model assumptions

Every model emphasizes some features of reality and leaves out others. This model assumes:

1. **The system contains both carts.** Looking at only one cart would not show conservation of the system's momentum.
2. **External impulse is negligible during the collision.** The track, Earth, friction, and other outside interactions do not significantly change the two-cart momentum during the short collision.
3. **Motion is one-dimensional.** The carts move only left or right.
4. **Masses remain constant.** Neither cart gains or loses material.
5. **The coefficient of restitution describes the collision.** The selected value represents how the carts separate.
6. **Friction before and after the collision is not modeled.**
7. **The input values are exact.** Measurement uncertainty is not included.
8. **The carts are represented as objects with velocities rather than as deforming materials.** Sound, heat, bending, and detailed contact forces are not directly modeled.
